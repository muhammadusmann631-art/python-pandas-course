### Data Frame
    A Pandas DataFrame is a two-dimensional table-like structure in Python where data is arranged in rows and columns. It’s one of the most commonly used tools for handling data and makes it easy to organize, analyze and manipulate data. It can store different types of data such as numbers, text and dates across its columns. The main parts of a DataFrame are:

1)Data: Actual values in the table.
2)Rows: Labels that identify each row.
3)Columns: Labels that define each data category.

 creat a  own data frame

### Adding garde using conditions
   -- List comprehentaion
   -- functions
   -- lambda

```python
data["grade"] = [
    "A+" if a >=90
    else "B"if a >=80
    else "C"if a >=65
    else "D"if a >=40
    else "F"
for a in data.percenatage]
```

```python
def persent_assing(grade):
    if grade >=90:
        return "A+"
    elif grade >=80:
        return "B"
    elif grade >= 65:
        return "C"
    elif grade >= 40:
        return "D"
    else:
        return "F"
```

```python
data["grade_1"]=data.percenatage.apply(persent_assing)
```
